## 3. ***Data Wrangling***

### Data Wrangling Code

````python
# Write your code to make your dataset analysis ready.

# --- 1. SETUP: DEFINE COLUMN LISTS ---
# These lists are now updated with the exact column names you provided.

datetime_cols = [
    'order_date_time',
    'Issue_reported at',
    'issue_responded',
    'Survey_response_Date'
]

# --- 2. DATA WRANGLING PIPELINE ---
print("--- Step 1: Starting Data Wrangling Pipeline ---")
# Make a copy to work on, keeping the original DataFrame safe
df_new = df.copy()

# Step 2.1: Handle Duplicate Rows
initial_rows = len(df_new)
df_new.drop_duplicates(inplace=True)
print(f"✅ Step 2.1: Removed {initial_rows - len(df_new)} duplicate rows.")

# Step 2.2: Correct Data Types (Object to Datetime)
for col in datetime_cols:
    df_new[col] = pd.to_datetime(df_new[col], errors='coerce')
print("✅ Step 2.2: Converted date-related columns to datetime format.")

# Step 2.3: Handle Missing Values
# Impute 'connected_handling_time' with its median
median_handling_time = df_new['connected_handling_time'].median()
df_new['connected_handling_time'].fillna(median_handling_time, inplace=True)

# Impute 'Customer_City' with its mode (most frequent value)
mode_customer_city = df_new['Customer_City'].mode()[0]
df_new['Customer_City'].fillna(mode_customer_city, inplace=True)
print("✅ Step 2.3: Handled missing values.")

# Step 2.4: Feature Engineering: Create a 'response_time_minutes' feature
df_new['response_time'] = df_new['issue_responded'] - df_new['Issue_reported at']
df_new['response_time_minutes'] = df_new['response_time'].dt.total_seconds() / 60
df_new['response_time_minutes'].fillna(df_new['response_time_minutes'].median(), inplace=True)
df_new.drop('response_time', axis=1, inplace=True)
print("✅ Step 2.4: Created new feature 'response_time_minutes'.")

# --- 3. FINAL VERIFICATION ---
print("\n--- Final Check: Dataset Info After Wrangling ---")
df_new.info()

print("\n--- Final Check: Missing Values After Wrangling ---")
print(df_new.isnull().sum())
````

### What all manipulations have you done and insights you found?

**Handle Duplicates:** Duplicate rows can bias your analysis and model training. The df.drop_duplicates(inplace=True) command finds and removes these identical rows permanently from our copied DataFrame.

**Correct Data Types:** Columns containing dates and times are often read as text (object). Converting them to a datetime format using pd.to_datetime is essential. It allows us to perform calculations, like finding the difference between two timestamps. errors='coerce' is a safety measure that turns any un-parseable date into NaT (Not a Time), which can then be handled as a missing value.

**Handle Missing Values:** We found missing values in our exploration. Instead of dropping rows, which can lead to data loss, we impute them:

For the numerical Connected handling time column, we fill missing values with the median. The median is used because it's less sensitive to extreme values (outliers) than the mean.

For the categorical Customer city column, we use the mode (the most frequently occurring city), which is the most logical replacement for a categorical feature.

**Feature Engineering:** This is a critical step where we create new, potentially more predictive variables from the existing data.

We created response_time_minutes by subtracting Issue reported at from Issue responded. A customer's satisfaction is very likely influenced by how quickly their issue is addressed. This new feature provides a direct, powerful signal to the model.

**Final Verification:** After performing these operations, it's good practice to run .info() and .isnull().sum() again to confirm that our changes have been applied correctly. We can see the new data types, the non-null counts, and the creation of our new feature.

## ***5. Hypothesis Testing***

### Based on your chart experiments, define three hypothetical statements from the dataset. In the next three questions, perform hypothesis testing to obtain final conclusion about the statements through your code and statistical testing.

**Hypothesis 1:** The Impact of Response Time on Satisfaction

Statement: "The average response time for dissatisfied customers (CSAT Score 1 or 2) is significantly longer than the average response time for satisfied customers (CSAT Score 4 or 5)."

Rationale: This is based on our findings in Charts 4, 12, and 13, which all suggested a strong negative relationship between wait time and satisfaction.

Statistical Test to be Used: Two-Sample Independent T-Test.

**Hypothesis 2:** The Effect of Service Channel on Satisfaction

Statement: "There is a significant difference in the average CSAT scores among the different customer service channels (e.g., Call, Chat, App, etc.)."

Rationale: Our analysis in Chart 2 showed that the 'Call' channel had a noticeably lower average CSAT score. This test will check if that difference is statistically significant.

Statistical Test to be Used: Analysis of Variance (ANOVA).

**Hypothesis 3:** The Value of Agent Experience

Statement: "The average CSAT score achieved by the most experienced agents is significantly higher than the average score achieved by the newest agents."

Rationale: In Chart 9, we observed a clear positive trend where more experienced agents received higher satisfaction scores. This test will validate that trend.

Statistical Test to be Used: Two-Sample Independent T-Test.

### Hypothetical Statement - 1

#### 1. State Your research hypothesis as a null hypothesis and alternate hypothesis.

Our goal is to determine if the average response time is significantly longer for dissatisfied customers compared to satisfied ones.

**Null Hypothesis(H0) :** The null hypothesis states that there is no difference in the mean response time between dissatisfied and satisfied customers. Any observed difference in our sample is due to random chance.

**Alternative Hypothesis(H1):** The alternative hypothesis is our research claim, based on the insights from our charts. It states that the mean response time for dissatisfied customers is significantly greater than that for satisfied customers.

#### 2. Perform an appropriate statistical test.

````python
# Perform Statistical Test to obtain P-Value

# Import the necessary library for statistical tests
from scipy import stats

# --- 1. Prepare the Data ---
# Create two separate groups based on CSAT score

# Group 1: Satisfied customers (CSAT Score is 4 or 5)
satisfied_group = df_new[df_new['CSAT Score'].isin([4, 5])]['response_time_minutes'].dropna()

# Group 2: Dissatisfied customers (CSAT Score is 1 or 2)
dissatisfied_group = df_new[df_new['CSAT Score'].isin([1, 2])]['response_time_minutes'].dropna()

# --- 2. Print Sample Means for Context ---
print(f"Average response time for SATISFIED customers: {satisfied_group.mean():.2f} minutes")
print(f"Average response time for DISSATISFIED customers: {dissatisfied_group.mean():.2f} minutes")
print("-" * 50)


# --- 3. Perform the T-Test ---
# We use a two-sample independent t-test.
# H₁ is that dissatisfied > satisfied, so we use alternative='greater'.
# We set equal_var=False to perform Welch's t-test, which doesn't assume equal variance.

t_statistic, p_value = stats.ttest_ind(
    dissatisfied_group,
    satisfied_group,
    equal_var=False,
    alternative='greater'
)

# --- 4. Display the Results ---
print(f"T-statistic: {t_statistic:.4f}")
print(f"P-value: {p_value:.4f}")
````

##### Which statistical test have you done to obtain P-Value?

More precisely, the test I conducted was Welch's T-test, which is a more robust version of the standard independent t-test. Here's a breakdown:

**Purpose:** This test is used to compare the means of two distinct, independent groups (in our case, satisfied_group and dissatisfied_group) to see if the difference between them is statistically significant or just due to random chance.

**Why Welch's T-test?:** I specifically used the Welch's variant by setting the parameter equal_var=False in the code. This is a better practice for real-world data because it does not assume the two groups have equal variances, making the result more reliable.

**Why One-Tailed?:** The test was also a one-tailed test (using alternative='greater'). This was done to specifically test our directional hypothesis (H1
) that the mean response time for dissatisfied customers is greater than the mean for satisfied customers, not just different.

##### Why did you choose the specific statistical test?

The choice was based on three key characteristics of our data and hypothesis:

**We Are Comparing Two Groups:** Our goal is to compare the average response_time_minutes between two distinct populations: dissatisfied customers (CSAT 1-2) and satisfied customers (CSAT 4-5). The t-test is the standard method for this.

**The Groups Are Independent:** A customer in the "satisfied" group is not in the "dissatisfied" group, and the observations in one group do not influence the observations in the other. This independence is a core requirement for this test.

**The Test is Robust (Welch's T-test):** By setting the parameter equal_var=False, we used Welch's T-test. This is a crucial detail. We don't have to assume that the spread (variance) of response times is the same for both satisfied and dissatisfied customers. Since it's safer to assume the variances are different in real-world data, Welch's test provides a more reliable and accurate p-value.

### Hypothetical Statement - 2

#### 1. State Your research hypothesis as a null hypothesis and alternate hypothesis.

**Null Hypothesis(H0):** The null hypothesis states that there is no difference in the mean response time between dissatisfied and satisfied customers. Any observed difference in our sample is due to random chance.

**Alternative Hypothesis(H1):** The alternative hypothesis is our research claim, based on the insights from our charts. It states that the mean response time for dissatisfied customers is significantly greater than that for satisfied customers.

#### 2. Perform an appropriate statistical test.

````python
# Perform Statistical Test to obtain P-Value

# Import the necessary library for statistical tests
from scipy import stats

# --- 1. Prepare the Data ---
# Find the unique customer service channels in the dataset
unique_channels = df_new['channel_name'].unique()

# Create a list of data samples, where each sample is the CSAT scores for one channel
channel_groups = []
for channel in unique_channels:
    channel_groups.append(df_new[df_new['channel_name'] == channel]['CSAT Score'].dropna())

# --- 2. Perform the ANOVA Test ---
# The f_oneway function takes each group as a separate argument.
# We use the * operator to unpack our list of groups into arguments.

f_statistic, p_value = stats.f_oneway(*channel_groups)


# --- 3. Display the Results ---
print(f"F-statistic: {f_statistic:.4f}")
print(f"P-value: {p_value:.4f}")
````

##### Which statistical test have you done to obtain P-Value?

**Purpose:** The ANOVA test is used to compare the means of three or more independent groups to determine if there is a statistically significant difference among them.

**Application:** In this case, our "groups" are the different customer service channels (e.g., 'Call', 'Chat', 'App'). The ANOVA test checks if the observed differences in their average CSAT Score are statistically significant or simply due to random variation in the data.

##### Why did you choose the specific statistical test?

**Comparing Multiple Groups:** Our hypothesis is about the CSAT Score across several different service channels ('Call', 'Chat', 'App', etc.). Since we have more than two groups to compare, a simple t-test is not appropriate.

**Limitations of a T-Test:** A t-test is specifically designed to compare the means of only two groups. While you could technically perform multiple t-tests (e.g., comparing Call vs. Chat, then Call vs. App, etc.), this is statistically incorrect.

**Avoiding Error Inflation:** Performing multiple t-tests on the same data dramatically increases the probability of a Type I error—that is, finding a statistically significant result by pure chance when no real difference exists.

**ANOVA's Solution:** The ANOVA test is the correct statistical method because it is designed to compare the means of all groups simultaneously in a single test. It determines if there is a significant difference somewhere among the groups without inflating the error rate, making it the right tool for this specific hypothesis.

**Alternative hypothesis(H1):** There is a significant different between value

#### 2. Perform an appropriate statistical test.

````python
# Perform Statistical Test to obtain P-Value

# Import the necessary library for statistical tests
from scipy import stats
import numpy as np

# --- 1. Prepare the Data ---
# To make the code robust, we'll first find the tenure buckets automatically.
# We assume the tenure buckets can be sorted logically (e.g., '0-3 Months', '12+ Months').
try:
    sorted_tenure_buckets = sorted(df_new['Tenure Bucket'].unique())
    newest_bucket = sorted_tenure_buckets[0]
    experienced_bucket = sorted_tenure_buckets[-1]

    print(f"Comparing newest agents ('{newest_bucket}') vs. most experienced agents ('{experienced_bucket}').")
    print("-" * 50)

    # Group 1: Most experienced agents
    experienced_agents_group = df_new[df_new['Tenure Bucket'] == experienced_bucket]['CSAT Score'].dropna()

    # Group 2: Newest agents
    newest_agents_group = df_new[df_new['Tenure Bucket'] == newest_bucket]['CSAT Score'].dropna()

    # --- 2. Print Sample Means for Context ---
    print(f"Average CSAT score for MOST EXPERIENCED agents: {experienced_agents_group.mean():.2f}")
    print(f"Average CSAT score for NEWEST agents: {newest_agents_group.mean():.2f}")
    print("-" * 50)


    # --- 3. Perform the T-Test ---
    # We test our hypothesis (experienced > newest) by using alternative='greater'.
    # We use Welch's T-test (equal_var=False) for robustness.

    t_statistic, p_value = stats.ttest_ind(
        experienced_agents_group,
        newest_agents_group,
        equal_var=False,
        alternative='greater'
    )

    # --- 4. Display the Results ---
    print(f"T-statistic: {t_statistic:.4f}")
    print(f"P-value: {p_value:.4f}")

except Exception as e:
    print(f"An error occurred. This might happen if the 'Tenure Bucket' values are not easily sortable. Error: {e}")
````

##### Which statistical test have you done to obtain P-Value?

**Purpose:** This test is the correct choice for comparing the means of two distinct, independent groups. Here, our groups are the newest agents and the most experienced agents.

**Why Welch's T-test?:** I used this specific variant by setting equal_var=False. This is the best practice as it provides a more accurate result by not assuming that the two groups have the same variance in their CSAT scores.

**Why One-Tailed?:** The test was a one-tailed test (using alternative='greater'). This was to specifically test our directional hypothesis that the mean CSAT score for the experienced agent group is greater than the mean score for the newest agent group.

##### Why did you choose the specific statistical test?

**We Are Comparing Two Groups:** Our hypothesis is designed to compare the average CSAT Score between two specific populations: the newest agents and the most experienced agents. The t-test is the standard method for comparing the means of exactly two groups.

**The Groups Are Independent:** The agents in the "newest" group are not the same as the agents in the "most experienced" group. Because the two sets of observations are unrelated, the samples are independent, which is a key requirement for this test.

**The Test is Robust (Welch's T-test):** By setting the equal_var=False parameter, we use Welch's T-test. This is the best practice because we don't have to assume that the performance variance is the same for new and experienced agents. This makes our conclusion more statistically reliable.

````python
# Remove Stopwords
# Import the NLTK library and download the stopwords list if you haven't already
import nltk
nltk.download('stopwords')
from nltk.corpus import stopwords

# --- 1. Prepare the Stop Words List ---
# Load the standard English stopwords into a set for fast lookup
stop_words = set(stopwords.words('english'))

# --- 2. Define the Removal Function ---
def remove_stopwords(text):
    # Ensure the input is a string
    if not isinstance(text, str):
        return ""
    # Split the text into words (tokenization)
    words = text.split()
    # Remove stopwords and join the words back into a string
    filtered_words = [word for word in words if word not in stop_words]
    return ' '.join(filtered_words)

# --- 3. Apply the Function ---
# We apply this function to our last processed column, 'remarks_cleaned'.
df_new['remarks_no_stopwords'] = df_new['remarks_cleaned'].apply(remove_stopwords)


# --- 4. Verification ---
print("--- Verifying Stop Word Removal ---")
# Display your actual data to see the changes
display(df_new[['remarks_cleaned', 'remarks_no_stopwords']].head())
````

````python
# Normalizing Text (i.e., Stemming, Lemmatization etc.)

# Import NLTK and download necessary resources
import nltk
from nltk.stem import WordNetLemmatizer
from nltk.corpus import wordnet
from nltk import pos_tag

# Download resources if you haven't already
nltk.download('wordnet')
nltk.download('omw-1.4')
nltk.download('averaged_perceptron_tagger')
nltk.download('averaged_perceptron_tagger_eng')
nltk.download('punkt_tab')


# --- 1. Create a function to map NLTK's POS tags to a format WordNetLemmatizer understands ---
def get_wordnet_pos(treebank_tag):
    if treebank_tag.startswith('J'):
        return wordnet.ADJ
    elif treebank_tag.startswith('V'):
        return wordnet.VERB
    elif treebank_tag.startswith('N'):
        return wordnet.NOUN
    elif treebank_tag.startswith('R'):
        return wordnet.ADV
    else:
        # Default to noun if the tag is not recognized
        return wordnet.NOUN

# --- 2. Create the Lemmatization Function ---
lemmatizer = WordNetLemmatizer()

def lemmatize_text(tokens):
    # Get the part-of-speech tag for each token
    pos_tagged_tokens = pos_tag(tokens)


  # Lemmatize each word with its corresponding POS tag

    lemmatized_tokens = [lemmatizer.lemmatize(word, get_wordnet_pos(pos)) for word, pos in pos_tagged_tokens]
    return lemmatized_tokens

# --- 3. Apply the Function ---
# We apply this function to our tokenized text column.
df_new['text_lemmatized'] = df_new['text_tokens'].apply(lemmatize_text)


# --- 4. Verification ---
print("--- Verifying Lemmatization ---")
# Display the tokenized and lemmatized lists side-by-side
display(df_new[['text_tokens', 'text_lemmatized']].head())
````

**Preserves Meaning:** Lemmatization uses a dictionary and contextual analysis (Part-of-Speech tagging) to reduce words to their actual base form, or lemma (e.g., "studies" becomes "study," and "better" becomes "good"). This preserves the semantic meaning of the text, which is crucial for understanding sentiment.

**Higher Accuracy:** Unlike Stemming, which is a crude, rule-based method that just chops off the ends of words (e.g., "studies" might become "studi"), lemmatization produces real, valid words. This leads to a cleaner, more interpretable set of tokens and a more accurate analysis.

**Better for Sentiment Analysis:** Because it correctly groups different forms of the same word (like "ran" and "running" to "run"), it allows the model to learn the sentiment associated with the core concept of the word more effectively.

Imbalanced dataset is relevant primarily in the context of supervised machine learning involving two or more classes.

Imbalance means that the number of data points available for different the classes is different: If there are two classes, then balanced data would mean 50% points for each of the class. For most machine learning techniques, little imbalance is not a problem. So, if there are 60% points for one class and 40% for the other class, it should not cause any significant performance degradation. Only when the class imbalance is high, e.g. 90% points for one class and 10% for the other, standard optimization criteria or performance measures may not be as effective and would need modification.

In our case the dataset dependent column data ratio is 85:15. So, during model creating it's obvios that there will be bias and having a great chance of predicting the majority one so frequently. SO the dataset should be balanced before it going for the model creation part.

````python
from imblearn.over_sampling import SMOTE

# Convert features to float32 (reduces memory usage)
X_train = X_train.astype('float32')

# Convert one-hot encoded labels back to a single column
if isinstance(y_train, pd.DataFrame):
    y_series = y_train.idxmax(axis=1).apply(lambda x: int(x.split('_')[1]))  # Extract numeric labels
else:
    y_series = y_train


# Get the count for each class before SMOTE
class_counts = y_series.value_counts()
majority_count = class_counts.max()
print("Majority class count:", majority_count)

# Create a sampling strategy that balances all classes to the majority count
smote_strategy = {cls: majority_count for cls in class_counts.index if cls != class_counts.idxmax()}

print("SMOTE strategy for balancing:", smote_strategy)

# Apply SMOTE with the balancing strategy
sm = SMOTE(sampling_strategy=smote_strategy, random_state=42)
X_train_resampled, y_train_resampled = sm.fit_resample(X_train, y_series)



# Print dataset info
print("\nFinal Dataset Sizes:")
print("X_train:", X_train_resampled.shape)
print("y_train:", y_train_resampled.shape)
print("X_test:", X_test.shape)
print("y_test:", y_test.shape)
````

# **Defining the ANN Model :**

````python
# Defining the learning scheduler

def scheduler(epoch, lr):
    if epoch < 10:
        return lr
    else:
        return lr * tf.math.exp(-0.1)
````

````python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, BatchNormalization, Dropout
from tensorflow.keras.regularizers import l2  # l2() is used for L2 regularization,which helps reduce overfitting in a neural network.

# Dropout rate
dropout_rate = 0.5

# Define the neural network model with BatchNormalization and Dropout layers
neural_classifier = Sequential(
    [
        Dense(128, activation="relu", kernel_regularizer=l2(),input_dim=X_train.shape[1]),
        BatchNormalization(),
        Dropout(dropout_rate),

        Dense(96, activation="relu", kernel_regularizer=l2()),
        BatchNormalization(),
        Dropout(dropout_rate),

        Dense(64, activation="relu", kernel_regularizer=l2()),
        BatchNormalization(),
        Dropout(dropout_rate),

        Dense(32, activation="relu", kernel_regularizer=l2()),
        BatchNormalization(),
        Dropout(dropout_rate),

        Dense(num_classes, activation="softmax")
    ]
)

# Print the model summary
neural_classifier.summary()
````
